```python
# 1. Import libraries 
import tensorflow as tf 
import matplotlib.pyplot as plt 
import numpy as np  
from tensorflow.keras.applications import VGG16 
from tensorflow.keras.applications.vgg16 import preprocess_input 
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense 
from tensorflow.keras.models import Model 
```

```python
dataset_path = r"C:\Users\Admin\Downloads\LP IV Updated Datasets\LP IV Updated Datasets\Object Detection\Intel Image Classification"
```

```python
# Load 80% of the dataset for training
# Resize every image to 64 x 64 pixels
# Load 32 images in one batch
# Use seed=42 to ensure the same dataset split every time

train_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    image_size=(64, 64),
    batch_size=32,
    validation_split=0.2,
    subset="training",
    seed=42
)

# Load the remaining 20% of the dataset for validation
# The same seed ensures that training and validation data do not overlap

test_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    image_size=(64, 64),
    batch_size=32,
    validation_split=0.2,
    subset="validation",
    seed=42
)

# Get the names of all the classes present in the dataset
class_names = train_data.class_names

# Display the class names
print("Classes:", class_names)
```

```python
# Define a function to preprocess images before passing them to VGG16
def preprocess(images, labels):

    # Apply VGG16-specific preprocessing
    # Converts RGB to BGR and subtracts ImageNet channel means
    images = preprocess_input(images)

    # Return the preprocessed images along with their labels
    return images, labels


# Apply preprocessing to the training dataset
train_data = train_data.map(preprocess)

# Apply preprocessing to the validation dataset
test_data = test_data.map(preprocess)
```

```python
# Load the VGG16 model pretrained on the ImageNet dataset
# weights="imagenet" loads previously learned weights
# include_top=False removes the original classification layers
# input_shape specifies the dimensions of our input images

base_model = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(64, 64, 3)
)
```

```python
# Freeze all layers of the pretrained VGG16 model
# Their weights will not change during initial training
# We use VGG16 as a feature extractor

base_model.trainable = False
```

```python
# Take the output feature maps from the pretrained VGG16 model
x = base_model.output

# Convert the feature maps into a single feature vector
# This reduces the spatial dimensions of the feature maps
x = GlobalAveragePooling2D()(x)

# Add a fully connected layer with 128 neurons
# ReLU activation introduces non-linearity
x = Dense(128, activation="relu")(x)

# Add the final output layer
# Number of neurons equals the number of classes
# Softmax predicts the probability of each class
output = Dense(
    len(class_names),
    activation="softmax"
)(x)

# Create the final model using VGG16 and our custom layers
model = Model(base_model.input, output)
```

```python
# Configure the model before training
model.compile(

    # Adam optimizer updates the model weights
    optimizer="adam",

    # Used for multiclass classification with integer labels
    loss="sparse_categorical_crossentropy",

    # Measure the percentage of correctly classified images
    metrics=["accuracy"]
)
```

```python
# Train the model using the training dataset
# epochs=2 means the model sees the complete training dataset twice
# Validation data is used to check performance after each epoch

history = model.fit(
    train_data,
    epochs=2,
    validation_data=test_data
)
```

```python
# Evaluate the trained model using the validation dataset
# This returns the loss and accuracy

loss, accuracy = model.evaluate(test_data)

# Display the classification accuracy
print("Test Accuracy:", accuracy)
```
