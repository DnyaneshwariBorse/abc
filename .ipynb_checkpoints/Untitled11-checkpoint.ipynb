```python
import re
from gensim.models import Word2Vec

with open("document3.txt", "r", encoding="utf-8") as file:
    document = file.read()

print(document)
```

```python
# Convert text to lowercase
document = document.lower()

# Remove punctuation and numbers
document = re.sub(r"[^a-z\s]", " ", document)

# Split document into sentences
sentence_list = re.split(r"[.!?]+", document)

# Tokenize each sentence
sentences = [sentence.split() for sentence in sentence_list if sentence.strip()]

print(sentences)
```

```python
cbow = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=5,
    min_count=1,
    sg=0,
    epochs=100
)
```

```python
print("Vocabulary:")
print(cbow.wv.index_to_key)
```

```python
word = "influenza"

if word in cbow.wv:
    print("Word:", word)
    print("Word vector:")
    print(cbow.wv[word])
else:
    print("Word not found")
```

```python
print(cbow.wv.most_similar("influenza", topn=5))
```

```python
print("Vocabulary size:", len(cbow.wv))
```
