```python
import pandas as pd
import re
from gensim.models import Word2Vec

data = pd.read_csv("twitter_dataset.csv")

data.head()
```

```python
def clean_tweet(tweet):
    tweet = str(tweet).lower()
    tweet = re.sub(r"http\S+|www\S+", "", tweet)
    tweet = re.sub(r"@\w+", "", tweet)
    tweet = re.sub(r"#", "", tweet)
    tweet = re.sub(r"[^a-z\s]", "", tweet)

    words = tweet.split()
    return words

sentences = data["Text"].apply(clean_tweet).tolist()

sentences[:5]
```

```python
cbow = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=5,
    min_count=2,
    sg=0,
    epochs=5
)
```

```python
print("Vocabulary size:", len(cbow.wv))
```

```python
word = list(cbow.wv.index_to_key)[0]

print("Word:", word)
print("Vector:", cbow.wv[word])
```

```python
cbow.wv.most_similar(word, topn=5)
```
